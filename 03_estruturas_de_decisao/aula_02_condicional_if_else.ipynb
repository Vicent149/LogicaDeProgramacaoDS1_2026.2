# 📓 Aula 02: Condicionais `if` e `else`

### 🎯 Objetivo da Aula
Aprender a estruturar fluxos de decisão simples e compostos utilizando a sintaxe `if` (se) e `else` (senão), respeitando a indentação obrigatória do Python.

---

## 1. Estrutura Básica

```python
if condicao:
    # Bloco executado se a condição for True (4 espaços de recuo)
    comando_1
    comando_2
else:
    # Bloco executado se a condição for False
    comando_3
```

> **Atenção:** Em Python, os dois pontos `:` no final da linha e o recuo (indentação) de 4 espaços definem o início e o fim de um bloco de código!


````python
# Exemplo do Professor: Sistema de Catraca
idade = int(input("Digite a idade do passageiro: "))

if idade >= 18:
    print("Acesso permitido: Tarifa adulta.")
else:
    print("Acesso com autorização: Tarifa estudante/menor.")
print("Operação finalizada.")
````
